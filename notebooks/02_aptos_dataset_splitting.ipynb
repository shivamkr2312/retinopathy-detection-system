```python

import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split

# Load labelled dataset
BASE = Path("../data/raw/aptos/aptos2019-blindness-detection")

df = pd.read_csv(BASE / "train.csv")

# First split: 70% training, 30% remaining
train_df, remaining_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["diagnosis"],
    random_state=42
)

# Second split: 15% validation, 15% internal testing
val_df, test_internal_df = train_test_split(
    remaining_df,
    test_size=0.50,
    stratify=remaining_df["diagnosis"],
    random_state=42
)

# Create output directory
output = Path("../data/processed/splits")
output.mkdir(parents=True, exist_ok=True)

# Save splits
train_df.to_csv(output / "train_split.csv", index=False)
val_df.to_csv(output / "validation_split.csv", index=False)
test_internal_df.to_csv(output / "internal_test_split.csv", index=False)

# Display results
print("TRAINING:", len(train_df))
print("VALIDATION:", len(val_df))
print("INTERNAL TEST:", len(test_internal_df))

print("\nClass distribution:")
print(pd.DataFrame({
    "Train": train_df["diagnosis"].value_counts().sort_index(),
    "Validation": val_df["diagnosis"].value_counts().sort_index(),
    "Internal Test": test_internal_df["diagnosis"].value_counts().sort_index()
}))

print("\nAll splits saved successfully!")
```
